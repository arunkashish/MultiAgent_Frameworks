```python
## PROMPT ENGINEERING -> providing instruction to an LLM Model how it should behave

from langchain_core.prompts import ChatPromptTemplate

#System -> Template is ther to define a prompt instruction in the LLM how it should behave, 
#User -> will be used as context to the LLM to search for

prompt=ChatPromptTemplate.from_messages(
    [
        ("system","your are an expert AI data scientist.Provide me answer in crisp and pointed clear"),  #system promt template
        ("user","{input}") 
    ]
)
prompt
```

```python
#CHAINING will combine model and prompt PROMPT|Model|Output -> should be the order

model1=ChatGroq(model="gemma2-9b-it")
chain=prompt|model1
#will go to prompt and check the user input and then go to LLM
response=chain.invoke({"input":"Tell me something about Langsmith"})
print(response.content)



```

```python
##OUTPUT PARCEL how the llm response should look like we can define it and we no need .content

from langchain_core.output_parsers import StrOutputParser

output_parse=StrOutputParser()

chain=prompt|model1|output_parse

#response1=chain.invoke({"input":"Tell me something about Langsmith"})
response1=chain.invoke({"input":"share the details of Realme P3"})

print(response1) #now the output we no need content to put in response1.content
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import PydanticOutputParser
from langchain_groq import ChatGroq
from pydantic import BaseModel,Field

model_gq1=ChatGroq(model="deepseek-r1-distill-llama-70b")


class Products(BaseModel):
    Product_Name: str = Field(description="Name of the product, including the model")
    Product_Detail: str = Field(description="Detailed specifications of the product, separated by pipes '|'")
    Price: float = Field(description="Price of the product, formatted as a USD amount")


pyd_parser=PydanticOutputParser(pydantic_object=Products)

prompt_chat=ChatPromptTemplate.from_messages(
    
        [
            ("system", "You are an E-Commerce Expert AI chatbot.Provide ONLY the information of the product explicitly mentioned by the user.Do not provide unrelated details.Your response MUST follow this JSON format exactly:\n{format_instr}"),
            ("human","{input}")
        ]
    
    
).partial(format_instr=pyd_parser.get_format_instructions())
#partial is used to inject static values into template variable
#in PromptTemplate need to give partial_variables inside Prmt template
#in ChatPromptTemplate need to do external inject .partial

chain_pydn=prompt_chat|model_gq1
response=chain_pydn.invoke({"input": "share the details of Realme P3"})
print(response.content)

```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import PydanticOutputParser
from langchain_groq import ChatGroq
from pydantic import BaseModel, Field

model_gq1 = ChatGroq(model="deepseek-r1-distill-llama-70b")

class Products(BaseModel):
    Product_Name: str = Field(description="Name of the product, including the model")
    Product_Detail: str = Field(description="Detailed specifications of the product, separated by pipes '|'")
    Price: float = Field(description="Price of the product, formatted as a USD amount",default=0)

pyd_parser = PydanticOutputParser(pydantic_object=Products)

prompt_chat = ChatPromptTemplate.from_messages([
    ("system", "You are an E-Commerce Expert AI chatbot. Provide ONLY the information of the product explicitly mentioned by the user. Do not provide unrelated details. Your response MUST follow this JSON format exactly:\n{format_instr}\n {input}"),
    ("user", "{input}")
]).partial(format_instr=pyd_parser.get_format_instructions())

chain_pydn = prompt_chat | model_gq1 | pyd_parser

response = chain_pydn.invoke({"input": "share the details of Realme P3"})

print(response)   
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
import re
import json
from langchain_core.documents import Document

# Load the PDF
pdf_doc = PyPDFLoader('/Users/arunkumar/Agentic_2.0/Langchain_May24/2-Langchain/temp.pdf')
loaded_pdf = pdf_doc.load()

# Input statement
statement = "Hello this is Arun from Singapore"

# Clean the statement and convert to Document
cleaned_statement = re.sub(r'[^a-zA-Z0-9\s]', '', statement).replace("\n", '').strip()
statement_doc = [Document(page_content=cleaned_statement)]

# Initialize text splitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=20)

# Split the PDF documents
final_documents = text_splitter.split_documents(loaded_pdf)

# Split the statement document
text_check = text_splitter.split_documents(statement_doc)

# Extract text chunks from final_documents and clean them
text_chunks = [re.sub(r'[^a-zA-Z0-9\s]', '', doc.page_content).replace("\n", '').strip() for doc in final_documents]

# Initialize embedding model (use one model for consistency, e.g., all-mpnet-base-v2 for 512 dimensions)
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")

# Generate embeddings for text chunks
embeddings = embedding_model.embed_documents(text_chunks)

# Generate embeddings for the statement
embeddings_stmt = embedding_model.embed_documents([cleaned_statement])

# Print embeddings for the statement
print("Statement Embedding:", embeddings_stmt[0])
print("Embedding length:", len(embeddings_stmt[0]))  # Should be 512

# Generate INSERT queries for the statement
for i, embedding in enumerate(embeddings_stmt):
    text = cleaned_statement  # Use the cleaned statement text
    # Format embedding as a JSON array for SQL
    embedding_str = "[" + ",".join(map(str, embedding)) + "]"
    #print(f"INSERT INTO VECTOR_IP (text, vector_data) VALUES ('{text}',JSON_ARRAY_PACK('{embedding_str}'));")
    #print(f"SELECT text, dot_product(vector_data, JSON_ARRAY_PACK('{embedding_str}')) AS score FROM VECTOR_IP ORDER BY score DESC LIMIT 10")

# Optionally, generate INSERT queries for PDF chunks
for i, (text, embedding) in enumerate(zip(text_chunks, embeddings)):
    embedding_str = "[" + ",".join(map(str, embedding)) + "]"
    # Escape single quotes in text to prevent SQL injection
    text = text.replace("'", "''")
    print(f"INSERT INTO VECTOR_IP (text, vector_data) VALUES ('{text}',JSON_ARRAY_PACK('{embedding_str}'));")




embedding_json = json.dumps(embeddings_stmt[0])
sql_query = f"""
SELECT statem, dot_product(vector_data, JSON_ARRAY_PACK('{embedding_json}')) AS score
FROM VECTOR_IP_T
ORDER BY score DESC
LIMIT 10
"""
print("SQL Query:", sql_query)
```
