```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
def load_data(file_name):
    try:
        data = pd.read_csv(file_name)
        return data
    except Exception as e:
        print(f"Failed to load {file_name}: {str(e)}")
        return None

# Prepare and plot data
def plot_gender_by_city(data):
    if data is not None:
        # Ensure columns are of suitable type
        data['Gender'] = data['Gender'].astype('category')
        data['City'] = data['City'].astype('category')

        # Group by City and Gender, then count
        gender_by_city = data.groupby(['City', 'Gender']).size().unstack('Gender').fillna(0)

        # Plot
        plt.figure(figsize=(10,6))
        gender_by_city.plot(kind='bar')
        plt.title('Distribution of Genders by City')
        plt.xlabel('City')
        plt.ylabel('Count')
        plt.legend(title='Gender')
        plt.xticks(rotation=90)  # For better readability
        plt.tight_layout()
        plt.show()

        # Additional text output for confirmation
        print("=== Execution Summary ===")
        print("Data Loaded:", bool(data.shape[0]))  
        print("Plot Generated:", True)  
        print("=== End of Summary ===")

        print("Task Completed.")

# Main Execution
if __name__ == "__main__":
    file_name ='/Users/arunkumar/Agentic_2.0/Project/Live-Class/Agentic 2.0/Autogen/7. Autogen Projects/2. Analyser GPT - Modular/temp/data.csv'
    data = load_data(file_name)
    plot_gender_by_city(data)
```
